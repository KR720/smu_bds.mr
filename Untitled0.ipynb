```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

INF = 1000
LEN = 100


# ==========================================
# 1. 벡터 및 행렬 연산
# ==========================================

def matrix_vector_mul(m, v):
    return np.dot(m, v)


def transform(p, m):
    hp = np.array([p[0], p[1], 1.0])
    q = matrix_vector_mul(m, hp)
    return np.array([q[0], q[1]])


# ==========================================
# 2. 이동 변환
# ==========================================

def translate(p, dx, dy):
    m = np.array([
        [1, 0, dx],
        [0, 1, dy],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 3. 회전 변환
# ==========================================

def rotate(p, theta):
    m = np.array([
        [np.cos(theta), -np.sin(theta), 0],
        [np.sin(theta),  np.cos(theta), 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 4. 크기 변환
# ==========================================

def scale(p, sx, sy):
    m = np.array([
        [sx, 0, 0],
        [0, sy, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 5. x축 방향 기울기
# ==========================================

def shear_x(p, sx):
    m = np.array([
        [1, sx, 0],
        [0, 1, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 6. y축 방향 기울기
# ==========================================

def shear_y(p, sy):
    m = np.array([
        [1, 0, 0],
        [sy, 1, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 7. x축 반전
# ==========================================

def reflect_x(p):
    m = np.array([
        [1, 0, 0],
        [0, -1, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 8. y축 반전
# ==========================================

def reflect_y(p):
    m = np.array([
        [-1, 0, 0],
        [0, 1, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 9. 원점 반전
# ==========================================

def reflect_origin(p):
    m = np.array([
        [-1, 0, 0],
        [0, -1, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 10. y = x 반전
# ==========================================

def reflect_yx(p):
    m = np.array([
        [0, 1, 0],
        [1, 0, 0],
        [0, 0, 1]
    ])
    return transform(p, m)


# ==========================================
# 11. 좌표축 그리기
# ==========================================

def draw_axis(ax):
    ax.axhline(0, color='black', alpha=0.3)
    ax.axvline(0, color='black', alpha=0.3)

    ax.set_xlim(-INF, INF)
    ax.set_ylim(-INF, INF)
    ax.set_aspect('equal')


# ==========================================
# 12. 사각형 그리기
# ==========================================

def draw_rectangle(ax, p1, p2, p3, p4):
    my_polygon = Polygon([p1, p2, p3, p4])
    ax.add_patch(my_polygon)


def apply_points(points, func, *args):
    return [func(p, *args) for p in points]


def draw_points(ax, points):
    draw_rectangle(ax, *points)


# ==========================================
# 13. 중앙 조종석
# ==========================================

def draw_cockpit(ax, p1, p2, p3, p4):

    points = [p1, p2, p3, p4]

    # Step 1. 원점으로 이동
    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    # Step 2. 크기 5배
    points = apply_points(
        points, scale, 5, 5
    )

    # Step 3. 45도 회전
    points = apply_points(
        points, rotate, np.pi/4
    )

    draw_points(ax, points)


# ==========================================
# 14. 왼쪽 날개축
# ==========================================

def draw_left_arm(ax, p1, p2, p3, p4):

    points = [p1, p2, p3, p4]

    # 사각형의 중심을 원점으로 이동
    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    # 가로로 길게 크기 변환
    points = apply_points(
        points, scale, 6, 0.7
    )

    # 왼쪽으로 이동
    points = apply_points(
        points, translate, -500, 0
    )

    draw_points(ax, points)


# ==========================================
# 15. 왼쪽 날개
# ==========================================

def draw_left_wing(ax, p1, p2, p3, p4):

    # 왼쪽 날개의 중심 기둥
    points = [p1, p2, p3, p4]

    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    points = apply_points(
        points, scale, 1.0, 8
    )

    points = apply_points(
        points, translate, -800, 0
    )

    draw_points(ax, points)

    # 왼쪽 위쪽 날개
    points = [p1, p2, p3, p4]

    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    points = apply_points(
        points, scale, 0.8, 2
    )

    points = apply_points(
        points, shear_x, 0.35
    )

    points = apply_points(
        points, translate, -820, 450
    )

    draw_points(ax, points)

    # 왼쪽 아래쪽 날개
    points = [p1, p2, p3, p4]

    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    points = apply_points(
        points, scale, 0.8, 2
    )

    points = apply_points(
        points, shear_x, -0.35
    )

    points = apply_points(
        points, translate, -820, -450
    )

    draw_points(ax, points)


# ==========================================
# 16. 오른쪽 날개축
# ==========================================

def draw_right_arm(ax, p1, p2, p3, p4):

    points = [p1, p2, p3, p4]

    points = apply_points(
        points, translate, -LEN/2, -LEN/2
    )

    points = apply_points(
        points, scale, 6, 0.7
    )

    points = apply_points(
        points, translate, -500, 0
    )

    points = apply_points(
        points, reflect_y
    )

    draw_points(ax, points)


# ==========================================
# 17. 오른쪽 날개
# ==========================================

def draw_right_wing(ax, p1, p2, p3, p4):

    # 오른쪽 날개는 왼쪽 날개의 y축 반전
    for cy, sh in [
        (0, 0),
        (450, 0.35),
        (-450, -0.35)
    ]:

        points = [p1, p2, p3, p4]

        points = apply_points(
            points, translate, -LEN/2, -LEN/2
        )

        if cy == 0:
            points = apply_points(
                points, scale, 1.0, 8
            )
            points = apply_points(
                points, translate, -800, 0
            )
        else:
            points = apply_points(
                points, scale, 0.8, 2
            )
            points = apply_points(
                points, shear_x, sh
            )
            points = apply_points(
                points, translate, -820, cy
            )

        points = apply_points(
            points, reflect_y
        )

        draw_points(ax, points)


# ==========================================
# 18. TIE Fighter 전체 그리기
# ==========================================

def draw_object(ax):

    p1 = np.array([0, 0])
    p2 = np.array([LEN, 0])
    p3 = np.array([LEN, LEN])
    p4 = np.array([0, LEN])

    # 1. 중앙 조종석
    draw_cockpit(ax, p1, p2, p3, p4)

    # 2. 왼쪽 날개축
    draw_left_arm(ax, p1, p2, p3, p4)

    # 3. 왼쪽 날개
    draw_left_wing(ax, p1, p2, p3, p4)

    # 4. 오른쪽 날개축
    draw_right_arm(ax, p1, p2, p3, p4)

    # 5. 오른쪽 날개
    draw_right_wing(ax, p1, p2, p3, p4)


# ==========================================
# 19. Main
# ==========================================

fig, ax = plt.subplots(figsize=(7, 7))

draw_axis(ax)
draw_object(ax)

plt.show()
```
